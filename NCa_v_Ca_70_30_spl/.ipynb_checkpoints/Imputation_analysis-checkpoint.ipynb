```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_row', None)

# load data
X_train = pd.read_csv('data/norm_X_train_NCa_Ca.csv')
X_test = pd.read_csv('data/norm_X_test_NCa_Ca.csv')
y_train = pd.read_csv('data/y_train_NCa_Ca.csv')
y_test = pd.read_csv('data/y_test_NCa_Ca.csv')


print(X_train.shape)
```

```python
Xy_NCa_Ca = pd.concat([X_train, y_train], axis=1)
```

```python
# file A
file_A = Xy_NCa_Ca.copy()
```

```python
# compare imputation for Ca in file_A


file_A2 = file_A[file_A["CACHEXSTAGE0VIG"] == 1]
```

```python
#file B

file_B = pd.read_csv('output_NCa_Ca.csv')
```

```python
# compare imputation for Ca in file_B


file_B2 = file_B[file_B["CACHEXSTAGE0VIG"] == 1]
```

```python


# mask of missing values in A
mask = file_A2.isna()

# values that were imputed
imputed_values = file_B2[mask]
```

```python
# print(file_B.isna().sum())
```

```python
comparison = []

for col in file_A2.columns:
    if file_A2[col].isna().sum() > 0:
        original = file_A2[col].dropna()
        imputed = file_B2[col][mask[col]]
        num_miss = len(imputed)
        print(imputed)
        
        comparison.append({
            'column': col,
            'Training_cohort_mean': original.mean(),
            'Imputed_data_mean': imputed.mean(),
            'Training_cohort_std': original.std(),
            'Imputed_data_std': imputed.std(),
            'Num_of_imputed_data': num_miss
        })

pd.DataFrame(comparison)
```
